
4. Copy the solution your partner wrote. 


```python
from typing import List

def move_zeros_to_end(nums: List[int]) -> List[int]:
    insert_pos = 0

    # Shift all non-zero elements forward
    for num in nums:
        if num != 0:
            nums[insert_pos] = num
            insert_pos += 1

    # Fill the remaining positions with zeros
    while insert_pos < len(nums):
        nums[insert_pos] = 0
        insert_pos += 1

    return nums

print(move_zeros_to_end([7, 7, 0, 0, 1, 3]))
```
